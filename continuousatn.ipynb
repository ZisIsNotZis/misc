```python
import numpy as np
def catn(q, k, v):
    qk = q@k.T
    qdk = qk[:,1:]-(qk:=qk[:,:-1])
    eqk = np.exp(qk)
    eqdk = np.exp(qdk)
    return (eqk/np.square(qdk))[:,None]/((eqk/qdk)[:,None]@(eqdk-1)[...,None])@((eqdk-1-qdk)[...,None]*v[:-1]+(eqdk*(qdk-1)+1)[...,None]*v[1:])
def atn(q, k, v):
    eqk = np.exp(q@k.T)
    return eqk/eqk.sum(1,keepdims=1)@v
```
