```python
from lzma import open


class Map:
    def __init__(s, f='log.xz'):
        s.f = f
        s.i = {}
        s.I = 0

        try:
            f = open(f, 'r')
            while _ := f.read(1):
                _ = f.read(_[0])
                s.i[_] = _ = int.from_bytes(f.read(2), 'little'), f.tell()
                f.seek(_[0], 1)
        except EOFError:
            pass
        s.I = f.tell()

    def __getitem__(s, i):
        if isinstance(i, str):
            i = i.encode()
        f = open(s.f, 'r')
        f.seek(s.i[i][1])
        return f.read(s.i[i][0])

    def __setitem__(s, i, a):
        if isinstance(i, str):
            i = i.encode()
        if isinstance(a, str):
            a = a.encode()
        print(s.F.write(b''.join([bytes([len(i)]), i, len(a).to_bytes(2, 'little'), a])))
        s.I += 1 + len(i) + 2
        s.i[i] = len(a), s.I
        s.I += len(a)
        s.F.flush()
        s.F.flush()


MAP = Map()


class Vec:
    def __init__(s, a=(), t='f', T='B', ondel=lambda _: None):
        s.T = T
        s.ondel = ondel
        if isinstance(a, bytes):
            k, b = np.frombuffer(a, t, 2)
            s.A = np.frombuffer(a, T, offset=k.itemsize * 2) * k + b
            s.i = len(s.A)
        else:
            s.A = np.pad(np.array(a, t), (0, 32))
            s.i = len(a)

    def __bytes__(s):
        b = s.a.min()
        a = s.a - b
        k = a.max() / np.iinfo(s.T).max
        return b''.join([k, b, (a / k).astype(s.T)])

    def __call__(s, a=None):
        if a is None:
            s.i = max(s.i - 1, 0)
            return s.A[s.i]
        if s.i == len(s.A):
            s.A.resize(s.i * 2)
        s.A[s.i] = a
        s.i += 1
        return s

    @property
    def a(s):
        return s.A[:s.i]

    def __del__(s):
        s.ondel(s)

    @staticmethod
    def alloc(i, A=MAP):
        return Vec(ondel=lambda a: a.i and A.__setitem__(i, bytes(a)))


MAP.i
```
